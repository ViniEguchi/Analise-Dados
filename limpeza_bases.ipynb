```python
def save_file(CAMINHO, df):
  return df.to_excel(CAMINHO, index=False)
```

```python
def resposta_1(df):
  df['Resposta 1'] = df['Resposta 1'].str.strip().str.upper().str.replace(' ', '').str.replace('-', '').str.replace('CINCO', '5')

  df['Resposta 1'] = df['Resposta 1'].str.replace('ADSB5', '5ADSB').str.replace('5ABSB', '5ADSB').str.replace('5ASDB', '5ADSB').str.replace('5ADB', '5ADSB')

  df.loc[df['Resposta 1'] != '5ADSB', 'Resposta 1'] = 'N/A'

  df['Resposta 1']
```

```python
def resposta_2(df):
  df['Resposta 2'] = pd.to_datetime(df['Resposta 2'], errors='coerce')

  df['Resposta 2']
```

```python
def resposta_3(df):
  df['Resposta 3'] = df['Resposta 3'].str.strip().str.upper()

  df.loc[df['Resposta 3'].str.contains('SÃO PAULO'), 'Resposta 3'] = 'SÃO PAULO'

  df.loc[df['Resposta 3'].str.contains('SP'), 'Resposta 3'] = 'SÃO PAULO'

  df.loc[df['Resposta 3'].str.contains('PAULISTA'), 'Resposta 3'] = 'PAULISTA'

  df.loc[df['Resposta 3'] == 'MATERNIDADE', 'Resposta 3'] = 'N/A'

  df['Resposta 3']
```

```python
def resposta_4(df):
  df['Resposta 4'] = df['Resposta 4'].str.strip().str.upper()

  df.loc[df['Resposta 4'].str.contains('STEFANINI'), 'Resposta 4'] = 'STEFANINI'

  df.loc[df['Resposta 4'].str.contains('EVERTEC'), 'Resposta 4'] = 'EVERTEC'

  df.loc[df['Resposta 4'].str.contains('MOTIVA'), 'Resposta 4'] = 'MOTIVA'

  df.loc[df['Resposta 4'].str.contains('DOCK'), 'Resposta 4'] = 'DOCK'

  df.loc[df['Resposta 4'].str.contains('SIMPRESS'), 'Resposta 4'] = 'SIMPRESS'

  df['Resposta 4']
```

```python
from os import error
def resposta_5(df):
  df['Resposta 5'] = df['Resposta 5'].astype(str)

  df['Resposta 5'] = df['Resposta 5'].str.replace('(incluindo eu)', '').str.replace('atualmente', '').str.replace(' ', '')

  df['Resposta 5'] = df['Resposta 5'].str.replace('Sóuma', '1').str.replace('Uma', '1')

  df['Resposta 5']
```

```python
def resposta_6(df):
  df['Resposta 6'] = df['Resposta 6'].astype(str)

  df.loc[df['Resposta 6'].str.upper().isin(['SIM', 'SI', 'OF COURSE', 'SIM, TRABALHEI SIM', 'SIM.']), 'Resposta 6'] = 'SIM'

  df.loc[df['Resposta 6'] != 'SIM', 'Resposta 6'] = 'NÃO'

  df['Resposta 6'].value_counts()
```

```python
CAMINHO1 = '/content/drive/MyDrive/Base_1.xlsx'

DF_LIMPO_1 = read_file(CAMINHO1)
resposta_1(DF_LIMPO_1)
resposta_2(DF_LIMPO_1)
resposta_3(DF_LIMPO_1)
resposta_4(DF_LIMPO_1)
resposta_5(DF_LIMPO_1)
resposta_6(DF_LIMPO_1)
```

```python
CAMINHO2 = '/content/drive/MyDrive/Base_2.xlsx'

DF_LIMPO_2 = read_file(CAMINHO2)
DF_LIMPO_2.columns = ['Numero Resposta', 'Resposta 1', 'Resposta 2', 'Resposta 3', 'Resposta 4', 'Resposta 5', 'Resposta 6']
resposta_1(DF_LIMPO_2)
resposta_2(DF_LIMPO_2)
resposta_3(DF_LIMPO_2)
resposta_4(DF_LIMPO_2)
resposta_5(DF_LIMPO_2)
resposta_6(DF_LIMPO_2)
```
