# 3) Criando Variáveis

Em Python, uma das primeiras coisas que aprendemos é **atribuir valores a um nome** → ou seja, guardar uma informação → variável.

→ É como criar uma pequena "etiqueta" que aponta para um valor.

- **Como fazemos isso?** → usamos o sinal de igual (`=`) para atribuir valores a uma variável:

```python
idade = 35
```

→ Aqui, a palavra `idade` passa a representar o número **35**. Dizemos que “atribuímos o valor 35 à variável `idade`”.

- **Outro exemplo:**

```python
nome = "Maria"
```

→ Agora, `nome` guarda o texto `"Maria"`.

---
Em Python, usamos **aspas** para definir **textos (strings)**.

→ Você pode usar **aspas simples** `'texto'` ou **aspas duplas** `"texto"` — ambas funcionam!

→ Atenção: Não misture os dois tipos no mesmo texto:

---

**Regras para nomear variáveis:**
- Começar com **letra**
- **Sem espaços** → use underline: `idade_paciente`
- **Não usar símbolos especiais** como `@`, `#`, `%`, etc.
- Python diferencia **maiúsculas de minúsculas** → `Idade` é diferente de `idade`

`````python
#criar variável de idade
idade = 90

#criar a variável nome
nome = "Clara"

#verificar as variáveies
print("idade:", idade)
print(f"Nome: {nome}")
`````

# 5) Outros Tipos de Variáveis

→ Até aqui, vimos variáveis que guardam números, textos e valores lógicos. Mas muitas vezes queremos guardar conjuntos de informações — por exemplo, uma lista de sintomas, nomes de pacientes ou resultados de exames.

→ Para isso, o Python oferece outras estruturas muito úteis. Vamos conhecer as principais:

<h3>Listas (list) </h3>

As listas são usadas para guardar vários valores em uma única variável.

É como uma “caixinha” onde podemos armazenar vários itens, organizados por ordem.

→ Como criar uma lista? Usamos colchetes [ ] e separamos os itens por vírgulas:

`````python
#criar a lista sintomas
sintomas = ['febre', 'tosse', 'dor de cabeça']

#verificar a lista sintomas
print(sintomas)

#ver o tipo da variavel sintomas
type(sintomas)
`````

`````python
#lista mescladas com int e str
lista_variada = ['oi', 1, 2, True]

print(lista_variada)
`````

Aqui, a variável sintomas guarda três valores.

→ Como acessar os itens? Cada item da lista tem uma posição (índice) que começa no zero:

`````python
# mostra o primeiro item: "febre"
print(sintomas[0])

# segundo item: "tosse"
print(sintomas[1])
`````

→ Como modificar a lista? Você pode adicionar, remover ou alterar itens:

`````python
# adiciona novo item ao final - append
sintomas.append('febre alta')

print(sintomas)

# remove o item "tosse" - remove
# sintomas.remove('tosse')

print(sintomas)

# altera o primeiro item
sintomas[0] = 'febre baixa'

print(sintomas)

#verifica como ficou a lista após as alterações
`````

---

<h3> Dicionários (dict) </h3>

Os dicionários são usados para guardar pares de informação → um nome (chave) e um valor.

É como uma ficha de cadastro: cada campo tem um nome e um valor correspondente.

→ Como criar um dicionário? Usamos chaves { } e separamos os pares com dois-pontos ":"

`````python
#criar o dicionário do paciente com nome, idade e se tem diabetes
dicionario = {
    'nome': 'Clara',
    'idade': 90,
    'tem_diabete': False
}

#verificar o tipo da variável paciente
type(dicionario)
`````

→ Como acessar as informações? Usamos o nome da chave:

`````python
#verificar o nome do paciente
print(dicionario['nome'])

#verificar se o paciente tem diabetes
print(dicionario['tem_diabete'])
`````

→ Como adicionar ou alterar?

`````python
#adicionar a informação peso do paciente
dicionario['peso'] = 70.0

#alterar o valor idade do paciente
dicionario['idade'] = 91

#verificar como ficou o dicionário
print(dicionario)
`````

Dicionários também são mutáveis.

---
<h3> Tuplas (tuple) </h3>

Parecidas com listas, mas não podem ser modificadas (são imutáveis).

```python
coordenadas = (23.5, 45.3)  # latitude e longitude
```

Úteis quando os dados não devem ser alterados, como datas ou posições fixas.


---

<h3> Conjuntos (set) </h3>
Usados para guardar valores únicos, sem repetições, e sem ordem específica.

`````python
#criar o set de doencas
doencas = {'gripe', 'dengue', 'covid', 'dengue'}

# verificar o set (duplicados sao removidos automaticamente)
print(doencas)
`````

Ótimos para eliminar duplicatas ou verificar interseções entre conjuntos.

| Tipo    | Estrutura  | Características principais           |
| ------- | ---------- | ------------------------------------ |
| `list`  | Lista      | Ordenada, mutável, aceita repetições |
| `dict`  | Dicionário | Pares chave\:valor, mutável          |
| `tuple` | Tupla      | Ordenada, **imutável**               |

# 7) Funções

À medida que nossos códigos crescem, é comum precisarmos repetir cálculos ou blocos de instruções. Em vez de escrever tudo de novo, podemos criar uma função.

Uma função é como uma “caixa de tarefas”: você dá uma entrada (informações), ela realiza um processo, e devolve uma saída (resultado).

Ela nos ajuda a:
- Organizar melhor o código;
- Evitar repetições;
- Facilitar a reutilização do que já foi feito.

---

<h3> Como criar uma função? </h3>

Usamos a palavra-chave def, escolhemos um nome para a função e definimos os parâmetros (informações que ela precisa receber):

```python
def nome_da_funcao(parametro1, parametro2):
    # instruções
    return resultado
```

---
<h3> Exemplo prático

`````python
def IMC_class(peso, altura):

  #calculo do IMC
  result = round(peso/(altura**2), 2)

  #definição da classe
  if result < 18.5:
    print(result, '- Baixo Peso.')
  elif result >= 18.5 and result < 25:
    print(result, '- Peso Normal')
  elif result >= 25 and result < 30:
    print(result, '- Excesso de peso.')
  elif result >= 30 and result < 35:
    print(result, '- Obesidade Grau I')
  elif result >= 35 and result < 40:
    print(result, '- Obesidade Grau II')
  elif result >= 40:
    print(result, '- Obesidade Grau III')
  else:
    print('Valor Inválido.')

  return result
`````

`````python
#Calcule o IMC de uma pessoa com 80 kg e 1.7m de altura
IMC_class(150, 1.7)
`````

## 8. Laços de repetição — `for` e `while`

Laços de repetição permitem executar um mesmo bloco de código várias vezes, evitando repetir comandos manualmente.

`````python
notas = [7.5, 8.0, 5.5, 9.0]

# Laço de repetição com cada iteração sendo sobre cada item da lista
for nota in notas:
    print(nota)
`````

`````python
# For que itera 5 vezes (range)
for i in range(5):
    print(i)

# O range possui: início, fim e passo
# For que começa em 2, vai até antes do 10, pulando de 2 em 2
for j in range(2, 10, 2):
    print(j)
`````

`````python
nomes = ["Ana", "Bruno", "Carlos"]

# enumerate() percorre a lista e retorna o índice e o valor
# i recebe o índice e nome recebe o conteúdo da posição
for i, nome in enumerate(nomes):
    print(i, nome)

# len(nomes) retorna a quantidade de elementos da lista
# range() gera os índices: 0, 1 e 2
# nomes[i] acessa o valor armazenado em cada posição
for i in range(len(nomes)):
    print(i, nomes[i])
`````

`````python
# Variável que controla a quantidade de tentativas
tentativas = 0

# Repete enquanto tentativas for menor que 3
while tentativas < 3:
    print(f'Tentativa: {tentativas + 1}')
    tentativas+=1
`````

## 12. Tratamento de erros
O tratamento de erros permite que o programa lide com situações inesperadas sem encerrar sua execução.

<h3>Principais operadores lógicos e relacionais</h3>

A tabela abaixo mostra o que cada operador faz e o resultado usando esses valores:

```python
a = 1  
b = 2
````
| Operador 🔣 | Significado 📝                            | Exemplo 💡               | Resultado 🎯 |
|-------------|--------------------------------------------|---------------------------|--------------|
| `==`        | Igual a                                    | `a == b`                  | `False`      |
| `!=`        | Diferente de                               | `a != b`                  | `True`       |
| `>`         | Maior que                                  | `a > b`                   | `False`      |
| `>=`        | Maior ou igual a                           | `a >= b`                  | `False`      |
| `<`         | Menor que                                  | `a < b`                   | `True`       |
| `<=`        | Menor ou igual a                           | `a <= b`                  | `True`       |
| `and`       | **E lógico** (ambas condições verdadeiras) | `a > 0 and b > 0`         | `True`       |
| `or`        | **Ou lógico** (uma condição verdadeira)    | `a > 0 or b < 0`          | `True`       |
| `not`       | **Negação** (inverte o valor lógico)       | `not (a == b)`            | `True`       |


`````python
notas = [
    [8,7,9],
    [5,6,7],
    [9,9,10]
]

medias = []

for nota in notas:
    m = 0
    for n in nota:
        m += n
    medias.append(m/3)

for i in range(len(medias)):
    print(f'Aluno {i+1}: média {medias[i]}')
`````
