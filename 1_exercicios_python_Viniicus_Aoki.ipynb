### QUESTÃO 30

> *Uma cantina universitária deseja controlar o saldo de créditos de um estudante. *

Crie um programa que inicie com saldo de R$ 50,00 e exiba repetidamente o menu: 

- **1** - Consultar saldo

- **2** - Adicionar créditos

- **3** - Realizar compra

- **0** - Sair. 

O usuário deve selecionar as opções por input. Ao realizar uma compra, o programa só deve descontar o valor se houver saldo suficiente; caso contrário, deve informar **`Saldo insuficiente`**. O menu deve continuar sendo exibido até que o usuário escolha 0.

```python
from IPython.display import clear_output
import time

def ex30():
    saldo = 50.0
    while 1 == 1:
        opcao = input(print("""
Favor escolher uma das opções abaixo:
    1 - Consultar saldo
    2 - Adicionar créditos
    3 - Realizar compra
    0 - Sair
        """))

        if int(opcao) == 0:
            break
        elif int(opcao) == 1:
            print(f'Saldo atual: R${saldo}')
        elif int(opcao) == 2:
            adicao = input(print('Digite a quantidade de creditos a serem adicionados: '))
            saldo += int(adicao)
        else:
            reducao = input(print('Digite o valor da compra: '))
            if saldo < int(reducao):
                print('Saldo insuficiente!')
            else:
                saldo -= int(reducao)
        time.sleep(3)
        clear_output(wait=True)

ex30()
```
